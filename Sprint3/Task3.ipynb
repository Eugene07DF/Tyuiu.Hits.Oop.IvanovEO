<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Продолжение проекта «Зоопарк». Базовый класс `Animal` (4 атрибута и 4 метода) и производные `Lion`, `Monkey`, `Penguin` (по 3 атрибута и 3-4 метода). Полиморфизм: переопределение `MakeSound`/`Move` и работа через `List<Animal>`. Перегрузка: `Eat()` (3 варианта), `Hunt`, `Climb`, `Swim`. Generic-класс `Enclosure<T> where T : Animal`.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

// ===== Базовый класс (4 атрибута и 4 метода) =====
public class Animal
{
    private string _name;
    private int _age;
    private double _weight;
    private int _energy = 50;

    public string Name   { get { return _name; }   set { _name = value; } }
    public int Age       { get { return _age; }    set { _age = value < 0 ? 0 : value; } }
    public double Weight { get { return _weight; } set { _weight = value; } }
    public int Energy    { get { return _energy; } protected set { _energy = Math.Max(0, Math.Min(100, value)); } }

    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    // Перекрытие (override) в производных классах
    public virtual void MakeSound() => Console.WriteLine($"{Name} издаёт звук");
    public virtual void Move() => Console.WriteLine($"{Name} движется");

    // Перегрузка (overload): три варианта Eat
    public void Eat()
    {
        Energy += 10;
        Console.WriteLine($"{Name} поел. Энергия: {Energy}");
    }

    public void Eat(string food)
    {
        Energy += 15;
        Console.WriteLine($"{Name} съел {food}. Энергия: {Energy}");
    }

    public void Eat(string food, int grams)
    {
        Energy += grams / 20;
        Weight += grams / 1000.0;
        Console.WriteLine($"{Name} съел {food} ({grams} г). Энергия: {Energy}, вес: {Weight:F2} кг");
    }

    public virtual void DisplayInfo() =>
        Console.WriteLine($"{GetType().Name} {Name}, {Age} лет, {Weight:F1} кг, энергия {Energy}");
}

// ===== Производные классы (по 3-4 атрибута и метода) =====
public class Lion : Animal
{
    public int PrideSize { get; set; } = 5;
    public int RoarPower { get; set; } = 80;
    public double ManeLength { get; set; } = 0.4;

    public Lion(string name, int age, double weight) : base(name, age, weight) { }

    public override void MakeSound() => Console.WriteLine($"{Name}: Рррр!");
    public override void Move() => Console.WriteLine($"{Name} движется мощными прыжками");

    // Перегрузка Hunt
    public void Hunt()
    {
        Energy -= 25;
        Console.WriteLine($"{Name} охотится. Энергия: {Energy}");
    }
    public void Hunt(string prey)
    {
        Energy -= 30;
        Console.WriteLine($"{Name} охотится на {prey}. Энергия: {Energy}");
    }

    public void Roar() => Console.WriteLine($"{Name} рычит с силой {RoarPower}");
    public void LeadPride() => Console.WriteLine($"{Name} ведёт прайд из {PrideSize} особей");
}

public class Monkey : Animal
{
    public int Intelligence { get; set; } = 7;
    public double TailLength { get; set; } = 0.6;
    public int Bananas { get; set; }

    public Monkey(string name, int age, double weight) : base(name, age, weight) { }

    public override void MakeSound() => Console.WriteLine($"{Name}: Уу-уу!");
    public override void Move() => Console.WriteLine($"{Name} прыгает по веткам");

    public void Climb() => Console.WriteLine($"{Name} лезет на дерево");
    public void Climb(int height) => Console.WriteLine($"{Name} лезет на высоту {height} м");
    public void CollectBanana() { Bananas++; Console.WriteLine($"{Name} собрал бананов: {Bananas}"); }
}

public class Penguin : Animal
{
    public double DiveDepth { get; set; } = 100;
    public int FeatherCount { get; set; } = 30000;
    public string Colony { get; set; } = "Антарктическая";

    public Penguin(string name, int age, double weight) : base(name, age, weight) { }

    public override void MakeSound() => Console.WriteLine($"{Name}: Кррря!");
    public override void Move() => Console.WriteLine($"{Name} переваливается по льду");

    public void Swim() => Console.WriteLine($"{Name} плывёт");
    public void Swim(double depth) => Console.WriteLine($"{Name} ныряет на {depth} м");
    public void Slide() => Console.WriteLine($"{Name} скользит на животе");
}

// ===== Generic-класс: вольер для животных определённого типа =====
public class Enclosure<T> where T : Animal
{
    private List<T> _animals = new List<T>();

    public string Title { get; set; }
    public int Capacity { get; set; }

    public Enclosure(string title, int capacity)
    {
        Title = title;
        Capacity = capacity;
    }

    public int Count => _animals.Count;

    public bool Add(T animal)
    {
        if (_animals.Count >= Capacity)
        {
            Console.WriteLine($"В вольере «{Title}» нет места для {animal.Name}");
            return false;
        }
        _animals.Add(animal);
        return true;
    }

    public void FeedAll(string food)
    {
        foreach (T a in _animals) a.Eat(food);
    }

    public void Show()
    {
        Console.WriteLine($"--- Вольер «{Title}» ({Count}/{Capacity}) ---");
        foreach (T a in _animals) a.DisplayInfo();
    }
}

public class Zookeeper
{
    public string Name { get; set; }
    public Zookeeper(string name) { Name = name; }

    // Полиморфизм: принимает любое животное
    public void Check(Animal a)
    {
        Console.WriteLine($"Смотритель {Name} проверяет {a.Name}");
        a.MakeSound();
        a.Move();
    }
}

// ===== Демонстрация =====
Zookeeper keeper = new Zookeeper("Иван");

Lion lion = new Lion("Симба", 6, 190);
Monkey monkey = new Monkey("Чита", 4, 25);
Penguin penguin = new Penguin("Пингви", 3, 15);

// Полиморфизм
List<Animal> zoo = new List<Animal> { lion, monkey, penguin };
foreach (Animal a in zoo) keeper.Check(a);

// Перегрузка
lion.Eat();
lion.Eat("мясо");
lion.Eat("мясо", 2000);
lion.Hunt();
lion.Hunt("антилопу");
monkey.Climb();
monkey.Climb(10);
penguin.Swim();
penguin.Swim(50);

// Generic
Enclosure<Lion> lions = new Enclosure<Lion>("Львы", 1);
lions.Add(lion);
lions.Add(new Lion("Нала", 5, 150));   // нет места
lions.FeedAll("мясо");
lions.Show();

Enclosure<Animal> mixed = new Enclosure<Animal>("Смешанный", 5);
mixed.Add(monkey);
mixed.Add(penguin);
mixed.Show();
```
