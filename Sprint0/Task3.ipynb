### Вариант задания №12: Система учета товаров (Item, SingleItem, PackageItem, SpecialItem)


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект демонстрирует применение принципов ООП (наследование и полиморфизм) на языке C#.

Реализован базовый класс `Item`, описывающий товар с атрибутами `ItemId`, `Name`, `Price` и методами `GetDetails()`, `CalculateDiscount()`, `ApplyDiscount(decimal discount)`.

На основе базового класса реализованы три производных класса:

- **SingleItem** (Единичный товар) — добавляет атрибут `UnitMeasure` (единица измерения) и переопределяет метод `GetDetails()`.
- **PackageItem** (Пакетный товар) — добавляет атрибут `QuantityPerPackage` (количество единиц в пакете) и переопределяет метод `CalculateDiscount()`.
- **SpecialItem** (Специальный товар) — добавляет атрибут `DiscountExpirationDate` (дата истечения скидки) и переопределяет метод `ApplyDiscount(decimal discount)`.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Item
{
    public int ItemId { get; set; }
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Item(int itemId, string name, decimal price)
    {
        ItemId = itemId;
        Name = name;
        Price = price;
    }

    public virtual string GetDetails()
    {
        return $"ID: {ItemId}, Название: {Name}, Цена: {Price:F2}";
    }

    public virtual decimal CalculateDiscount()
    {
        return Price * 0.05m;
    }

    public virtual void ApplyDiscount(decimal discount)
    {
        Price -= discount;
        if (Price < 0)
        {
            Price = 0;
        }
    }
}

public class SingleItem : Item
{
    public string UnitMeasure { get; set; }

    public SingleItem(int itemId, string name, decimal price, string unitMeasure)
        : base(itemId, name, price)
    {
        UnitMeasure = unitMeasure;
    }

    public override string GetDetails()
    {
        return $"{base.GetDetails()}, Единица измерения: {UnitMeasure}";
    }
}

public class PackageItem : Item
{
    public int QuantityPerPackage { get; set; }

    public PackageItem(int itemId, string name, decimal price, int quantityPerPackage)
        : base(itemId, name, price)
    {
        QuantityPerPackage = quantityPerPackage;
    }

    public override decimal CalculateDiscount()
    {
        decimal baseDiscount = base.CalculateDiscount();
        decimal quantityBonus = QuantityPerPackage * 0.5m;
        return baseDiscount + quantityBonus;
    }

    public override string GetDetails()
    {
        return $"{base.GetDetails()}, Количество единиц в пакете: {QuantityPerPackage}";
    }
}

public class SpecialItem : Item
{
    public DateTime DiscountExpirationDate { get; set; }

    public SpecialItem(int itemId, string name, decimal price, DateTime discountExpirationDate)
        : base(itemId, name, price)
    {
        DiscountExpirationDate = discountExpirationDate;
    }

    public override void ApplyDiscount(decimal discount)
    {
        base.ApplyDiscount(discount);
        Console.WriteLine($"Скидка {discount:F2} действует до {DiscountExpirationDate:d}");
    }

    public override string GetDetails()
    {
        return $"{base.GetDetails()}, Скидка действует до: {DiscountExpirationDate:d}";
    }
}

List<Item> items = new List<Item>
{
    new SingleItem(1, "Ноутбук", 55000m, "шт"),
    new PackageItem(2, "Батарейки", 300m, 4),
    new SpecialItem(3, "Смартфон", 40000m, DateTime.Now.AddDays(7))
};

foreach (Item item in items)
{
    Console.WriteLine(item.GetDetails());
    decimal discount = item.CalculateDiscount();
    Console.WriteLine($"Рассчитанная скидка: {discount:F2}");
    item.ApplyDiscount(discount);
    Console.WriteLine($"Цена после скидки: {item.Price:F2}");
    Console.WriteLine(new string('-', 40));
}
```
