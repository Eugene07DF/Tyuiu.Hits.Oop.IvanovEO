### Вариант задания 

Система «Зоопарк»: иерархия животных, смотритель и зоопарк.


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Продолжение проекта «Зоопарк». Базовый класс `Animal` (4 атрибута и 4 метода). Простое наследование: `Animal -> Mammal`. Сложное (многоуровневое): `Animal -> Mammal -> Lion` и `Animal -> Bird -> Duck`. Множественное через интерфейсы: `Duck : Bird, IFlyable, ISwimmable`.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public interface IFlyable { void Fly(); }
public interface ISwimmable { void Swim(); }

// ===== Базовый класс (4 атрибута, 4 метода) =====
public class Animal
{
    private string _name;
    private int _age;
    private double _weight;
    private int _energy = 50;

    public string Name   { get { return _name; }   set { _name = value; } }
    public int Age       { get { return _age; }    set { _age = value < 0 ? 0 : value; } }
    public double Weight { get { return _weight; } set { _weight = value; } }
    public int Energy    { get { return _energy; } protected set { _energy = Math.Max(0, Math.Min(100, value)); } }

    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    public virtual void MakeSound() => Console.WriteLine($"{Name} издаёт звук");
    public void Eat(int food)
    {
        Energy += food;
        Weight += food * 0.01;
        Console.WriteLine($"{Name} поел. Энергия: {Energy}");
    }
    public void Sleep()
    {
        Energy += 30;
        Console.WriteLine($"{Name} поспал. Энергия: {Energy}");
    }
    public virtual void DisplayInfo() =>
        Console.WriteLine($"{GetType().Name} {Name}, {Age} лет, {Weight:F1} кг, энергия {Energy}");
}

// ===== ПРОСТОЕ наследование: Animal -> Mammal =====
public class Mammal : Animal
{
    public int BodyTemperature { get; set; }
    public bool HasFur { get; set; }
    public int Litter { get; set; }
    public string Diet { get; set; }

    public Mammal(string name, int age, double weight, bool hasFur) : base(name, age, weight)
    {
        HasFur = hasFur;
        BodyTemperature = 37;
        Diet = "Всеядное";
    }

    public void Nurse() => Console.WriteLine($"{Name} кормит детёнышей");
    public void RegulateTemperature() => Console.WriteLine($"{Name} поддерживает {BodyTemperature}°C");
    public void GiveBirth(int count)
    {
        Litter = count;
        Console.WriteLine($"{Name} родил(а) детёнышей: {count}");
    }
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Мех: {HasFur}, питание: {Diet}");
    }
}

// ===== СЛОЖНОЕ (многоуровневое) наследование: Animal -> Mammal -> Lion =====
public class Lion : Mammal
{
    public int PrideSize { get; set; }
    public double ManeLength { get; set; }
    public int RoarPower { get; set; }
    public bool IsAlpha { get; set; }

    public Lion(string name, int age, double weight) : base(name, age, weight, true)
    {
        Diet = "Хищник";
        PrideSize = 5;
        ManeLength = 0.4;
        RoarPower = 80;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Рррр!");
    public void Roar() => Console.WriteLine($"{Name} рычит с силой {RoarPower}");
    public void Hunt()
    {
        Energy -= 25;
        Console.WriteLine($"{Name} охотится. Энергия: {Energy}");
    }
    public void LeadPride()
    {
        IsAlpha = true;
        Console.WriteLine($"{Name} ведёт прайд из {PrideSize} особей");
    }
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  Прайд: {PrideSize}, грива: {ManeLength} м");
    }
}

public class Monkey : Mammal
{
    public int Intelligence { get; set; } = 7;
    public double TailLength { get; set; } = 0.6;
    public int Bananas { get; set; }
    public bool CanUseTools { get; set; } = true;

    public Monkey(string name, int age, double weight) : base(name, age, weight, true) { }

    public override void MakeSound() => Console.WriteLine($"{Name}: Уу-уу!");
    public void Climb() => Console.WriteLine($"{Name} лезет на дерево");
    public void CollectBanana() { Bananas++; Console.WriteLine($"{Name} собрал бананов: {Bananas}"); }
    public void UseTool() => Console.WriteLine($"{Name} использует инструмент");
}

// ===== Птицы: Animal -> Bird -> Duck (+ МНОЖЕСТВЕННОЕ через интерфейсы) =====
public class Bird : Animal
{
    public double Wingspan { get; set; }
    public string FeatherColor { get; set; }
    public bool LaysEggs { get; set; } = true;
    public int EggCount { get; set; }

    public Bird(string name, int age, double weight, double wingspan) : base(name, age, weight)
    {
        Wingspan = wingspan;
        FeatherColor = "Серый";
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Чик-чирик!");
    public void LayEggs(int n) { EggCount += n; Console.WriteLine($"{Name} снесла яиц: {n}"); }
    public void Preen() => Console.WriteLine($"{Name} чистит перья");
    public void BuildNest() => Console.WriteLine($"{Name} строит гнездо");
}

// Один базовый класс + два интерфейса = множественное наследование в C#
public class Duck : Bird, IFlyable, ISwimmable
{
    public bool WebbedFeet { get; set; } = true;
    public double DiveDepth { get; set; } = 1.5;
    public int MigrationKm { get; set; } = 2000;
    public string Habitat { get; set; } = "Пруд";

    public Duck(string name, int age, double weight) : base(name, age, weight, 0.9)
    {
        FeatherColor = "Коричневый";
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Кря!");
    public void Fly() => Console.WriteLine($"{Name} летит ({MigrationKm} км миграции)");
    public void Swim() => Console.WriteLine($"{Name} плавает, ныряет до {DiveDepth} м");
    public void Dive() { Energy -= 10; Console.WriteLine($"{Name} ныряет. Энергия: {Energy}"); }
    public void Migrate() => Console.WriteLine($"{Name} улетает на юг на {MigrationKm} км");
}

public class Zookeeper
{
    public string Name { get; set; }
    public Zookeeper(string name) { Name = name; }

    public void Feed(Animal a, int food)
    {
        Console.WriteLine($"Смотритель {Name} кормит {a.Name}");
        a.Eat(food);
    }
}

// ===== Демонстрация =====
Zookeeper keeper = new Zookeeper("Иван");
Lion lion = new Lion("Симба", 6, 190);
Monkey monkey = new Monkey("Чита", 4, 25);
Duck duck = new Duck("Дональд", 2, 1.2);

List<Animal> zoo = new List<Animal> { lion, monkey, duck };
foreach (Animal a in zoo)
{
    a.DisplayInfo();
    a.MakeSound();
    keeper.Feed(a, 20);
}

lion.Hunt();
lion.LeadPride();
lion.Nurse();           // метод Mammal (простое наследование)
monkey.Climb();
duck.Fly();             // интерфейс IFlyable
duck.Swim();            // интерфейс ISwimmable
duck.LayEggs(6);        // метод Bird
duck.Sleep();           // метод Animal (сложное наследование: Animal -> Bird -> Duck)
```
