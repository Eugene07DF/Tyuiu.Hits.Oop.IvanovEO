Зоопарк

### Вариант задания 

Система «Зоопарк»: животные, смотритель и сам зоопарк.


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект моделирует зоопарк. Базовый класс `Animal` и производные `Lion`, `Monkey`; класс `Zookeeper` (смотритель) кормит животных; класс `Zoo` хранит список животных и смотрителя и координирует их взаимодействие.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сеттеров и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Animal
{
    private string _name;
    private int _hunger; // 0..100

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Hunger
    {
        get { return _hunger; }
        set { _hunger = Math.Max(0, Math.Min(100, value)); }
    }

    public Animal() : this("Без имени", 50) { }

    public Animal(string name, int hunger)
    {
        Name = name;
        Hunger = hunger;
    }

    public virtual void MakeSound() => Console.WriteLine($"{Name} издаёт звук");

    public void Feed(int food) => Hunger -= food;
}

public class Lion : Animal
{
    public Lion(string name) : base(name, 70) { }
    public override void MakeSound() => Console.WriteLine($"{Name}: Рррр!");
}

public class Monkey : Animal
{
    public Monkey(string name) : base(name, 40) { }
    public override void MakeSound() => Console.WriteLine($"{Name}: Уу-уу!");
}

public class Zookeeper
{
    private string _name;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public Zookeeper(string name)
    {
        Name = name;
    }

    // Взаимодействие объектов: смотритель кормит животное
    public void FeedAnimal(Animal animal, int food)
    {
        animal.Feed(food);
        Console.WriteLine($"{Name} покормил {animal.Name}. Голод: {animal.Hunger}");
    }
}

public class Zoo
{
    private List<Animal> _animals = new List<Animal>();
    private Zookeeper _keeper;

    public string Title { get; set; }

    public Zookeeper Keeper
    {
        get { return _keeper; }
        set { _keeper = value; }
    }

    public Zoo(string title, Zookeeper keeper)
    {
        Title = title;
        Keeper = keeper;
    }

    public void AddAnimal(Animal animal) => _animals.Add(animal);

    // Зоопарк поручает смотрителю накормить всех
    public void FeedAll(int food)
    {
        foreach (Animal a in _animals)
            Keeper.FeedAnimal(a, food);
    }

    public void Voice()
    {
        foreach (Animal a in _animals)
            a.MakeSound();
    }
}

Zookeeper keeper = new Zookeeper("Иван");
Zoo zoo = new Zoo("Городской зоопарк", keeper);
zoo.AddAnimal(new Lion("Симба"));
zoo.AddAnimal(new Monkey("Чита"));
zoo.Voice();
zoo.FeedAll(30);
```
